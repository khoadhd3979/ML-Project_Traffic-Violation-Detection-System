# Mục mới

```python
import yaml, hashlib, re, collections
from pathlib import Path

YAML = Path("/content/dataset/content/smart-helmet-detection-using-dl")
cfg = yaml.safe_load(YAML.read_text())
names = cfg["names"]; names = dict(enumerate(names)) if isinstance(names, list) else names
print("nc:", cfg.get("nc"), "| names:", names)
root = Path(cfg.get("path", YAML.parent))

def split_dirs(key):
    v = cfg.get(key)
    if v is None: return []
    return [(root / p) if not Path(p).is_absolute() else Path(p) for p in ([v] if isinstance(v, str) else v)]

def label_dir(img_dir):
    s = str(img_dir)
    return Path(s.replace("/images", "/labels")) if "/images" in s else img_dir.parent / "labels"

stats = {}
for split in ("train", "val"):
    inst, imgs = collections.Counter(), collections.Counter()
    empty = malformed = n_img = missing_lbl = 0
    stems, hashes = [], {}
    for d in split_dirs(split):
        for im in sorted(d.rglob("*")):
            if im.suffix.lower() not in {".jpg", ".jpeg", ".png", ".bmp", ".webp"}: continue
            n_img += 1; stems.append(im.stem)
            hashes[hashlib.md5(im.read_bytes()).hexdigest()] = im.name
            lp = label_dir(im.parent) / (im.stem + ".txt")
            if not lp.exists(): missing_lbl += 1; continue
            lines = [l for l in lp.read_text().splitlines() if l.strip()]
            if not lines: empty += 1; continue
            seen = set()
            for l in lines:
                p = l.split()
                try:
                    c = int(float(p[0])); vals = list(map(float, p[1:]))
                    ok = c in names and len(vals) >= 4 and all(0 <= v <= 1 for v in vals)
                except Exception:
                    ok = False
                if not ok: malformed += 1; continue
                inst[c] += 1; seen.add(c)
            for c in seen: imgs[c] += 1
    stats[split] = dict(n_img=n_img, inst=inst, imgs=imgs, empty=empty, malformed=malformed,
                        missing_lbl=missing_lbl, stems=stems, hashes=hashes)

for split, s in stats.items():
    print(f"\n== {split}: {s['n_img']} ảnh | label rỗng {s['empty']} | thiếu label {s['missing_lbl']} | dòng lỗi {s['malformed']}")
    print(f"{'class':32s}{'instances':>10s}{'images':>8s}")
    for c in sorted(names):
        print(f"{c} {names[c]:30s}{s['inst'][c]:>10d}{s['imgs'][c]:>8d}")
    tot = sum(s["inst"].values()) or 1
    mx, mn = max(s["inst"].values(), default=0), min((s["inst"][c] for c in names), default=0)
    print("imbalance max/min:", round(mx / mn, 1) if mn else "có class 0 instance")

def src(stem):
    s = re.sub(r"\.rf\.[0-9a-f]+$", "", stem); s = re.sub(r"_(jpg|png|jpeg)$", "", s)
    return re.sub(r"[-_ ]?\d+$", "", s)
dup = set(stats["train"]["hashes"]) & set(stats["val"]["hashes"])
print("\nẢnh trùng hệt giữa train/val:", len(dup))
st, sv = {src(x) for x in stats["train"]["stems"]}, {src(x) for x in stats["val"]["stems"]}
print("Nguồn (heuristic theo tên file) xuất hiện ở cả hai split:", len(st & sv), "/", len(sv))
```
